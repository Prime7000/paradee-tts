```python
from paradee import Paradee, SAMPLE_RATE
import soundfile as sf
from IPython.display import Audio, display


tts = Paradee()   # downloads the 9 MB model from the Hugging Face Hub the first time
```

```python
%%time
text = """
        A quantum computer is a special type of computer that uses the rules of quantum physics to process information.
Normal computers use bits, which are either 0 or 1.
Quantum computers use quantum bits called qubits. A qubit can behave like 0, 1, or a combination of both at the same time.

       """

audio = tts(text)
sf.write("audio.wav", audio, SAMPLE_RATE)
display(Audio(filename="audio.wav"))
```
